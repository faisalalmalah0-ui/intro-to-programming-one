```python
def filter_longer_than(strings, k):
    result = []
    for s in strings:
        if len(s) > k:
            result.append(s)
    return result
my_list = ["cake", "pizza", "banana"]
k = 3
print(filter_longer_than(my_list, k))
```

```python
PI = 3.14
def area(shape, *dims):
    shape = shape.lower()
    if shape == "circle" and len(dims) == 1:
        return "Circle Area: " + str(PI * dims[0] ** 2)
    elif shape == "rectangle" and len(dims) == 2:
        return "Rectangle Area: " + str(dims[0] * dims[1])
    return "Invalid shape or dimensions"

print(area("circle", 7))
print(area("rectangle", 8, 5))
```

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = {}
        self.customer_orders = []
    def add_item_to_menu(self, name, price):
        self.menu_items[name] = price
    def book_tables(self, table_number, customer_name):
        if table_number in self.book_table:
            print("This table is already booked")
        else:
            self.book_table[table_number] = customer_name
            print("Table booked")
    def customer_order(self, customer_name, items):
        total = 0
        ordered = []
        for item in items:
            if item in self.menu_items:
                ordered.append(item)
                total += self.menu_items[item]
            else:
                print(item, "is not available")
        self.customer_orders.append({"customer": customer_name, "items": ordered, "total": total})
    def print_menu(self):
        print("Menu")
        for name, price in self.menu_items.items():
            print(name, ":", price)
    def print_reservations(self):
        print("Reservations")
        for table, customer in self.book_table.items():
            print(table, ":", customer)
    def print_orders(self):
        print("Orders")
        for order in self.customer_orders:
            print(order["customer"], ":", ", ".join(order["items"]), "Total:", order["total"])
r = Restaurant()
n = int(input("Enter number of items: "))
for i in range(n):
    name = input("Item name: ")
    price = float(input("Item price: "))
    r.add_item_to_menu(name, price)
n = int(input("Enter number of reservations: "))
for i in range(n):
    table_number = int(input("Table number: "))
    customer_name = input("Customer name: ")
    r.book_tables(table_number, customer_name)
n = int(input("Enter number of orders: "))
for i in range(n):
    customer_name = input("Customer name: ")
    items = input("Enter items: ").split()
    r.customer_order(customer_name, items)
r.print_menu()
r.print_reservations()
r.print_orders()
```

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def get_name(self):
        return self.name
    def set_name(self, name):
        self.name = name
    def get_age(self):
        return self.age
    def set_age(self, age):
        self.age = age
class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def get_name(self):
        return self.name
    def set_name(self, name):
        self.name = name
    def get_price(self):
        return self.price
    def set_price(self, price):
        self.price = price
class MainBikeRental:
    HOURLY_RATE = 40
    def __init__(self, bikes=None, rentals=None):
        self.bikes = bikes if bikes is not None else []
        self.rentals = rentals if rentals is not None else []
    def get_bikes(self):
        return self.bikes
    def set_bikes(self, bikes):
        self.bikes = bikes
    def get_rentals(self):
        return self.rentals
    def set_rentals(self, rentals):
        self.rentals = rentals
    def addBike(self, bike):
        self.bikes.append(bike)
    def requestBike(self, customer, num_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6")
            return False
        if num_of_bikes <= 0:
            print("Enter a valid number")
            return False
        if num_of_bikes <= len(self.bikes):
            rented = [self.bikes.pop() for _ in range(num_of_bikes)]
            self.rentals.append({"customer": customer, "bikes": rented})
            print(customer.get_name(), "rented", num_of_bikes, "bike(s)")
            return True
        print("Not enough bikes")
        return False
    def returnBike(self, customer, rentalTime, num_of_bikes):
        for rental in self.rentals:
            if rental["customer"] is customer and len(rental["bikes"]) >= num_of_bikes:
                returned = rental["bikes"][:num_of_bikes]
                rental["bikes"] = rental["bikes"][num_of_bikes:]
                self.bikes.extend(returned)
                if not rental["bikes"]:
                    self.rentals.remove(rental)
                bill = rentalTime * self.HOURLY_RATE
                print(customer.get_name(), "returned", num_of_bikes, "bike(s)")
                print("Total:", bill)
                return bill
        print("No rental found")
        return 0
    def totalCost(self):
        return sum(b.get_price() for b in self.bikes)
```
