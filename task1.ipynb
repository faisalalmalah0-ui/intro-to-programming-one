```python
s = input("Enter a string: ")
chars = []
nums = []
 
for c in s:
        if (c >= "a" and c <= "z") or (c >= "A" and c <= "Z"):
            chars.append(c)
        elif c >= "0" and c <= "9":
            nums.append(int(c))
 
word = "".join(chars)
 
total = 0
for n in nums:
        total = total + n
 
print('Word: "' + word + '", Sum: ' + str(total))
```

```python
pw = input("Enter a password: ")
specials = ["@", "#", "$"]
has_upper = False
has_special = False

for c in pw:
    if c >= "A" and c <= "Z":
        has_upper = True
    if c in specials:
        has_special = True
if len(pw) >= 8 and len(pw) <= 16 and has_upper and has_special:
    print("Password is Secure.")
else:
    print("Password is not valid.")
```

```python
students={
    "faisal": [95, 90, 92],
    "walied": [100, 95, 98],
    "ali": [80, 85, 70],
    "ziad": [99, 85, 85],
    "moaz": [96, 93, 85],
    "wael": [90, 95, 70],
    "mai": [82, 85, 77],
    "akram": [90, 75, 85],
}
for name in students.keys():
    scores = students[name]

    total = 0
    for s in scores:
        total = total + s

    avg = total / 3
    shown = int(avg * 10) / 10    

    result = name + " Average: " + str(shown)
    if avg  > 90:
        result = result + " - Honor Roll"
    print(result)
```

```python
parts = input("Enter numbers separated by spaces: ").split()
numbers = []
for p in parts:
    numbers.append(int(p))
distinct = []
for x in numbers:
    if x not in distinct:
        distinct.append(x)
result = []
for u in distinct:
    count = 0
    for x in numbers:
        if x == u:
            count = count + 1
    result.append((u, count))

print("result:",result)
```

```python
n = int(input("Enter n: ")) #الفاكتوريل

fact = 1
total = 0
for i in range(1, n + 1):
    fact = fact * i
    total = total + i

if n >= 1 and fact % total == 0:
    print("Factorial is " + str(fact) + ", Sum is " + str(total) + ". " + str(fact) + " is divisible by " + str(total) + ".")
else:
    print("Factorial is " + str(fact) + ", Sum is " + str(total) + ". " + str(fact) + " is not divisible by " + str(total) + ".")
```

```python
store = {"laptop": 1000.0, "iphone": 800.0, "airpods": 150.0}
cart = []

store = {"laptop": 1000.0, "iphone": 800.0, "airpods": 150.0}
cart = []
print("Items:", store)

while True:
    action = input("Add, Remove or Done? ").lower()

    if action == "done":
        break
    name = input("Item name: ").lower()

    if action == "add":
        if name in store:
            cart.append(name)
        else:
            print("Error: item not in the store")
    elif action == "remove":
        if name in cart:
            cart.remove(name)
        else:
            print("Error: item not in your cart")
    else:
        print("Error: type Add, Remove or Done")
total = 0
for item in cart:
    total = total + store[item]
print("Total Bill: $" + str(total))
```

```python
parts = input("Enter numbers separated by spaces: ").split()
numbers = []
for p in parts:
    numbers.append(int(p))

peaks = []
for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
```

```python
d = {"A": 1, "B": 2, "C": 1}
flipped = {}
for key in d.keys():
    value = d[key]
    if value in flipped:
        flipped[value].append(key)
    else:
        flipped[value] = [key]
print(flipped)
```

```python
text = input("Enter a string: ").lower()
vowels = ["a", "e", "i", "o", "u"]
result = {}
for i in range(len(text)):
    ch = text[i]
    if ch in vowels:
        if ch in result:
            result[ch].append(i)
        else:
            result[ch] = [i]

print(result)
```
