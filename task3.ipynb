```python
import numpy as np

a = np.random.randint(1, 101, size=(3, 3))
normalized = (a - a.min()) / (a.max() - a.min())
print("Original:\n", a)
print("Normalized:\n", normalized)
```

```python
import numpy as np

arr = np.random.randint(1, 101, size=10)
print("Array:", arr)
x = float(input("Enter a value X: "))
closest = arr[np.abs(arr - x).argmin()]
print("Closest value to", x, "is:", closest)
```

```python
import numpy as np

m = np.random.rand(5, 5)
print("Array:\n", m)
second_largest = np.sort(m, axis=0)[-2]
print("Second-largest per column:", second_largest)
```

```python
import numpy as np

mat = np.random.randint(1, 100, size=(6, 4))
print("Matrix:\n", mat)
row_sums = mat.sum(axis=1)
print("Row sums:", row_sums)
even_rows = np.where(row_sums % 2 == 0)[0]
print("Indices of rows with even sum:", even_rows)
```

```python
import numpy as np

v = np.random.randint(1, 51, size=30)
print("Original:", v)
v[(v % 3 == 0) | (v % 5 == 0)] = -1
print("Result:  ", v)
```

```python
import numpy as np

r = np.random.rand(3, 3)
print("Original:\n", r)

centered = r - r.mean(axis=1, keepdims=True)
print("Row-centered:\n", centered)
example = np.array([[10, 20, 30], [5, 5, 5], [0, 10, 20]])
print("Example output:\n", example - example.mean(axis=1, keepdims=True))
```
